```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
import statsmodels.api as sm
import matplotlib.pyplot as plt
import statsmodels.api as sm
import numpy as np
import scipy.stats as stats
import random
from scipy.stats import norm
from plotnine import ggplot
from plotnine import ggplot, aes, geom_point, geom_pointrange, facet_wrap, labs, position_dodge
```

```python
url = (
    "https://raw.githubusercontent.com/ranalytics/"
    "r-tutorials/master/Edition_2015/Data/RIKZ.txt"
)

rikz = pd.read_csv(url, sep="\t")

print(rikz.shape)
print(rikz.columns)
print(rikz.head())
print(len(rikz))
```

The data, which can be accessed via the Data4Ecologists library, were collected at 5 stations at each of 9 beaches. The beaches were located at 3 different levels of exposure to waves (high, medium, or low). For now, we will focus on the following variables:

- Richness = species richness (i.e., the number of species counted).
- NAP = height of the sample site (relative to sea level). Lower values indicate sites closer to or further below sea level, which implies more time spent submerged.
- Beach = a unique identifier for the specific beach where the observation occurred

```python
X = rikz[['NAP']]
y = rikz[['Richness']]

X = sm.add_constant(X)
model = sm.OLS(y,X).fit()
predicted = model.fittedvalues
plt.figure(figsize=(12,8))
plt.scatter(X['NAP'], y, color='black')
plt.plot(X['NAP'], predicted, color='blue')
plt.grid(color='lightgray', linestyle='-', linewidth=1)
plt.ylabel('Richness')
plt.xlabel('NAP')
plt.show()
```

```python
model.summary()
```

```python
residuals = model.resid
fitted = model.fittedvalues
```

```python

plt.figure(figsize=(10,8))
plt.scatter(fitted, residuals)
plt.ylabel('Residuals')
plt.xlabel('Predicted')
plt.title('Residual plot')
plt.grid(color='lightgray', linestyle='-', linewidth=1)
plt.axhline(0, linewidth=1, c='b')
plt.show()
```

__Homogeneity of variance__: the plot of residuals versus fitted values shows a clear increase in variance as we move from left to right (i.e., from smaller to larger predicted values).

```python
plt.figure(figsize=(10,8))
plt.grid(color='lightgray', linestyle='-', linewidth=1)
stats.probplot(residuals, dist="norm", plot=plt)
plt.gca().get_lines()[0].set_markerfacecolor('black')       # Scatter points
plt.gca().get_lines()[0].set_markeredgecolor('black')   # Scatter points
plt.gca().get_lines()[1].set_color('blue')  
plt.show()
```

In a Q–Q plot, each point compares:

- x-position: the residual value expected under a normal distribution
- y-position: the corresponding residual actually observed

The blue line represents approximate agreement with normality.

Position relative to the line:
- Point on the line: the observed residual is close to what normality predicts.
- Point above the line: the residual is more positive—or less negative—than expected.
- Point below the line: the residual is more negative—or less positive—than expected.

The location along the plot also matters:

- Left side: smallest, most negative residuals
- Middle: typical residuals near the centre
- Right side: largest, most positive residuals

We will calculate a 95% confidence interval for the intercept and slope parameters using a bootstrap, which we will then compare to standard t-based confidence intervals for these parameters. To generate the bootstrap confidence interval, we will:

- Generate a bootstrap data set by resampling our original observations with replacement.
- Calculate our bootstrap statistics, $\hat{\beta_0}$ and $\hat{\beta_1}$, by fitting a linear model to our bootstrap data set.
We will repeat steps 1 and 2 a large number of times to form our bootstrap distribution, which we then plot:


Randomly sample beaches with replacement, keeping the original number of beaches.
- Whenever a beach is selected, include all five observations from that beach.
- If a beach is selected twice, include its five observations twice; another beach may not appear at all.
- Fit the statistical model to this bootstrap dataset.
- Repeat many times to estimate uncertainty, such as standard errors and confidence intervals.

The bootstrap distributions are relatively bell-shaped and symmetric, so we could use a Normal approximation to generate confidence intervals for the intercept and slope parameters. To do so, we would calculate the bootstrap standard error as the standard deviation of the bootstrap statistics. We would then take our original sample statistics  ±

 . Alternatively, we could use percentiles of the bootstrap distribution to calculate a confidence interval. We demonstrate these two approaches, below, and compare them to the standard t-based interval.

It is standard error. The confusing part is that a standard error is calculated as a standard deviation—but of a particular distribution.

- Standard deviation (SD): spread of individual data values.
- Standard error (SE): standard deviation of an estimate across repeated samples.

Definition is always the same:

Standard error=standard deviation of an estimator’s sampling distribution

What changes between models and statistics is:

Which estimator you are studying: mean, slope, proportion, median, etc.
How its sampling distribution is obtained or approximated.
Which formula or computational method calculates its SE.

```python
# Fit the model to the original data to get our estimates -> fitted above

# Calculate bootstrap standard errors
boot_std = [np.std(boot_rikz['const']), np.std(boot_rikz['NAP'])]
confidence_int = model.conf_int(alpha=0.05)
print(f'Standard error : {boot_std}')
print(f'Condifence intervals: {confidence_int}')

# bootstrap normal -> using original, not bootstrapped values, just the bootstrap error

boot_normal = [
    [ model.params['const'] - (1.96*boot_std[0]), model.params['const'] + (1.96*boot_std[0])],
    [ model.params['NAP'] - (1.96*boot_std[1]),   model.params['NAP'] + (1.96*boot_std[1])]
]

print(f'\n\nNormal distribution :\nconst -> {boot_normal[0]},\nNAP -> {boot_normal[1]}')


# bootstrap percentile

quantiles = [ np.quantile(boot_rikz['const'], [0.025, 0.975]) , np.quantile(boot_rikz['NAP'], [0.025, 0.975])]

print(f'\nQuantiles for:\nconst (intercept) -> lower: {quantiles[0][0]}, upper: {quantiles[0][1]},\nNAP (slope) -> lower: {quantiles[1][0]}, upper: {quantiles[1][1]}') 
```

```python
conf_data = confdats = np.vstack([
    confidence_int,
    boot_normal,
    quantiles
])
conf_data
```

```python
conf_data = pd.DataFrame(conf_data, columns=["LCL", "UCL"])
```

```python
conf_data["method"] = np.repeat(
    ["t-based", "bootstrap-Normal", "bootstrap-percentile"],
    2
)
```

```python
conf_data["parameter"] = np.tile(
    ["Intercept", "Slope"],
    3
)

conf_data["estimate"] = np.tile(model.params, 3)
```

```python
parameters = ["Intercept", "Slope"]
methods = conf_data["method"].unique()

fig, axes = plt.subplots(1, 2, figsize=(20, 8))
for ax, parameter in zip(axes, parameters):
    subset = conf_data[conf_data["parameter"] == parameter]

    for x, (_, row) in enumerate(subset.iterrows()):
        lower_error = row["estimate"] - row["LCL"]
        upper_error = row["UCL"] - row["estimate"]
        ax.grid(color='lightgray', linestyle='-', linewidth=1)

        ax.errorbar(
            x,
            row["estimate"],
            yerr=[[lower_error], [upper_error]],
            fmt="o",
            capsize=5,
            label=row["method"]
        )

    ax.set_title(parameter)
    ax.set_xticks(range(len(subset)))
    ax.set_xticklabels(subset["method"], rotation=20)
    ax.set_ylabel("Estimate")
plt.tight_layout()
plt.show()
```

Comparison of t-based and bootstrap intervals for the intercept and slope of the regression relating the NAP to the Richness of the rikz data

!!! This plot is not necessary for performing the bootstrap.

The actual bootstrap consists of:

1. Resampling the observations with replacement.
2. Fitting the model to every bootstrap sample.
3. Saving the coefficients.
4. Using those coefficients to calculate standard errors or confidence intervals.

The plot is only a visual comparison of:

- t-based intervals
- bootstrap-normal intervals
- bootstrap-percentile intervals

We can bootstrap and report the results without creating this plot. However, it is useful for checking whether the different methods produce similar intervals.
